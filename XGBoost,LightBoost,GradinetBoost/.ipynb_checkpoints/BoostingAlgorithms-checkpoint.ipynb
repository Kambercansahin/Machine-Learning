```python
fill_list=["capacity","occu_perc"]
df[fill_list]=df[fill_list].fillna(df[fill_list].median())
```

```python
df.isnull().sum()
```

```python
df.drop("date",axis=1,inplace=True)
```

```python
df.head()
```

```python
sns.pairplot(df)
plt.show()
```

```python
#dependent independent features
X=df.drop("ticket_price",axis=1)
y=df["ticket_price"]
```

```python
#train test split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=15)
```

```python
#standart scaller is not necessary for boosting but I use all models 
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)

#Metrics
from sklearn.metrics import r2_score
```

```python
#Linear regression
from sklearn.linear_model import LinearRegression
lin_reg=LinearRegression()
lin_reg.fit(X_train,y_train)
y_pred_lin_train=lin_reg.predict(X_train)# we are undertanding if model has overfitting 
y_pred_lin_test=lin_reg.predict(X_test)
print("Score Train",r2_score(y_train,y_pred_lin_train))
print("Score Test",r2_score(y_test,y_pred_lin_test))
```

```python
#polynomial Regression
from sklearn.preprocessing import PolynomialFeatures
poly=PolynomialFeatures(degree=3,include_bias=True)
X_train_poly=poly.fit_transform(X_train)
X_test_poly=poly.transform(X_test)
poly_regression=LinearRegression()
poly_regression.fit(X_train_poly,y_train)

y_pred_poly_train=poly_regression.predict(X_train_poly)# we are undertanding if model has overfitting 
y_pred_poly_test=poly_regression.predict(X_test_poly)
print("Score Train",r2_score(y_train,y_pred_poly_train))
print("Score Test",r2_score(y_test,y_pred_poly_test))
print("-------------------------------------")
poly=PolynomialFeatures(degree=4,include_bias=True)
X_train_poly=poly.fit_transform(X_train)
X_test_poly=poly.transform(X_test)
poly_regression=LinearRegression()
poly_regression.fit(X_train_poly,y_train)

y_pred_poly_train=poly_regression.predict(X_train_poly)# we are undertanding if model has overfitting 
y_pred_poly_test=poly_regression.predict(X_test_poly)
print("Score Train",r2_score(y_train,y_pred_poly_train))
print("Score Test",r2_score(y_test,y_pred_poly_test))
```

```python
#KNN
from sklearn.neighbors import KNeighborsRegressor
knn=KNeighborsRegressor()
knn.fit(X_train,y_train)

y_pred_knn_train=knn.predict(X_train)
y_pred_knn_test=knn.predict(X_test)

print("Score Train",r2_score(y_train,y_pred_knn_train))
print("Score Test",r2_score(y_test,y_pred_knn_test))
```

```python
#Decision Tree
from sklearn.tree import DecisionTreeRegressor
tree=DecisionTreeRegressor()
tree.fit(X_train,y_train)

y_pred_tree_train=tree.predict(X_train)
y_pred_tree_test=tree.predict(X_test)

print("Score Train",r2_score(y_train,y_pred_tree_train))
print("Score Test",r2_score(y_test,y_pred_tree_test))
```

```python
#RandomForrest
from sklearn.ensemble import RandomForestRegressor
random_tree=RandomForestRegressor()
random_tree.fit(X_train,y_train)

y_pred_random_train=random_tree.predict(X_train)
y_pred_random_test=random_tree.predict(X_test)

print("Score Train",r2_score(y_train,y_pred_random_train))
print("Score Test",r2_score(y_test,y_pred_random_test))
```

```python
#Adaboosting
from sklearn.ensemble import AdaBoostRegressor,GradientBoostingRegressor
adaboost=AdaBoostRegressor()
adaboost.fit(X_train,y_train)

y_pred_adaboost_train=adaboost.predict(X_train)
y_pred_adaboost_test=adaboost.predict(X_test)

print("Score Train",r2_score(y_train,y_pred_adaboost_train))
print("Score Test",r2_score(y_test,y_pred_adaboost_test))
```

```python
#gradient Boosting
gbr=GradientBoostingRegressor()
gbr.fit(X_train,y_train)

y_pred_gbr_train=gbr.predict(X_train)
y_pred_gbr_test=gbr.predict(X_test)

print("Score Train",r2_score(y_train,y_pred_gbr_train))
print("Score Test",r2_score(y_test,y_pred_gbr_test))
```

```python
#XGBoost
from xgboost import XGBRegressor
xgb=XGBRegressor()
xgb.fit(X_train,y_train)
y_pred_xgb_train=xgb.predict(X_train)
y_pred_xgb_test=xgb.predict(X_test)

print("Score Train",r2_score(y_train,y_pred_xgb_train))
print("Score Test",r2_score(y_test,y_pred_xgb_test))
```

```python
#Lightboost
from lightgbm import LGBMRegressor
lgb=LGBMRegressor()
lgb.fit(X_train,y_train)

y_pred_lgb_train=lgb.predict(X_train)
y_pred_lgb_test=lgb.predict(X_test)

print("Score Train",r2_score(y_train,y_pred_lgb_train))
print("Score Test",r2_score(y_test,y_pred_lgb_test))
```
