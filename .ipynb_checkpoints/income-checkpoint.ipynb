```python

list_nan=["workclass","occupation","native-country"]
for i in list_nan:
    df[i]=df[i].replace("?",np.nan)
```

```python
for i in list_nan:
    df[i]=df[i].fillna(df[i].mode()[0])
```

```python
for i in list_nan:
    print(df[i].value_counts())
    print()
```

```python
sns.pairplot(df,hue="income")
```

```python
#Dependent independent features
X=df.drop("income",axis=1)
y=df["income"]
```

```python
#Train test split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=15)
```

```python
X_train.isnull().sum()
```

```python
categoric_list=['workclass',
 'marital-status',
 'occupation',
 'relationship',
 'race',
 'sex',
 'native-country']
```

```python
X_train[categoric_list].nunique()
```

```python

#natice -country a lot of unique values so i don't use one hot for native-country
#Calculated the average income (>50K) rate for each native-country to use as a numeric feature instead of one-hot encoding.
y_train_binary=y_train.apply(lambda x : 1 if x.strip()==">50K" else 0)
y_train_binary.groupby(X_train['native-country']).mean()
```

```python
target=y_train_binary.groupby(X_train['native-country']).mean()
```

```python
X_train['native_country'] = X_train['native-country'].map(target)
X_train['native_country'] = X_train['native_country'].fillna(y_train_binary.mean())

X_test['native_country'] = X_test['native-country'].map(target)
X_test['native_country'] = X_test['native_country'].fillna(y_train_binary.mean())
```

```python

X_train.head()
```

```python
X_train.drop("native-country",axis=1,inplace=True)
X_test.drop("native-country",axis=1,inplace=True)
```

```python
one_hot=['workclass',
 'marital-status',
 'occupation',
 'relationship',
 'race',
 'sex']
```

```python
from sklearn.preprocessing import OneHotEncoder,LabelEncoder
from sklearn.compose import ColumnTransformer

encoder=ColumnTransformer(transformers=[("one-hot",OneHotEncoder(handle_unknown="ignore",sparse_output=False),one_hot)],remainder="passthrough")
encode_label=LabelEncoder()
X_train= encoder.fit_transform(X_train)
X_test= encoder.transform(X_test)
y_train=encode_label.fit_transform(y_train)
y_test=encode_label.transform(y_test)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python

from sklearn.linear_model import LogisticRegression
linear=LogisticRegression()
linear.fit(X_train,y_train)
y_pred_linear=linear.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
print(accuracy_score(y_test,y_pred_linear))
print(classification_report(y_test,y_pred_linear))
print(confusion_matrix(y_test,y_pred_linear))
```

```python
#KNN
from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier()
knn.fit(X_train,y_train)
y_pred_knn=knn.predict(X_test)
print(accuracy_score(y_test,y_pred_knn))
print(classification_report(y_test,y_pred_knn))
print(confusion_matrix(y_test,y_pred_knn))
```

```python
#NavieBayes
from sklearn.naive_bayes import GaussianNB
gausian=GaussianNB()
gausian.fit(X_train,y_train)
y_pred_gausian=gausian.predict(X_test)
print(accuracy_score(y_test,y_pred_gausian))
print(classification_report(y_test,y_pred_gausian))
print(confusion_matrix(y_test,y_pred_gausian))
```

```python
#SVC
from sklearn.svm import SVC
svc=SVC()
svc.fit(X_train,y_train)
y_pred_svc=svc.predict(X_test)
print(accuracy_score(y_test,y_pred_svc))
print(classification_report(y_test,y_pred_svc))
print(confusion_matrix(y_test,y_pred_svc))
```

```python
#Decision Tree
from sklearn.tree import DecisionTreeClassifier
tree=DecisionTreeClassifier()
tree.fit(X_train,y_train)
y_pred_tree=tree.predict(X_test)
print(accuracy_score(y_test,y_pred_tree))
print(classification_report(y_test,y_pred_tree))
print(confusion_matrix(y_test,y_pred_tree))
```

```python
#RandomForest
from sklearn.ensemble import RandomForestClassifier
randomforest=RandomForestClassifier()
randomforest.fit(X_train,y_train)
y_pred_random=randomforest.predict(X_test)
print(accuracy_score(y_test,y_pred_random))
print(classification_report(y_test,y_pred_random))
print(confusion_matrix(y_test,y_pred_random))
```

```python
#Hyperparameters Tunning
from sklearn.model_selection import GridSearchCV,RandomizedSearchCV
#for KNN
params_knn={"n_neighbors":[2,3,4,5,6,10,15,20],
            "weights":["uniform","distance"],
            "algorithm":["auto","ball_tree","kd_tree","brute"]}


#for SVC 
params_svc={"C":[1,10,20,30,50,100],
            "kernel":["linear","poly","rbf","sigmoid"],
            "gamma":["scale","auto",1,2,3]}


#for decision Tree
params_decision={"criterion":["gini","entropy","log_loss"],
                 "splitter":["best","random"],
                 "max_depth":[2,3,4,5,6,7,8,10,16,None],
                 "max_features":[2,3,4,5,6,"sqrt","log2"],
                 "min_samples_split":[1,2,3,4,5,6,7,8,10]}

#for random forest
params_random_forest={"max_depth": [5, 8, 15, None, 10],
                      "max_features": [5, 7, "auto", 8],
                      "min_samples_split": [2, 8, 15, 20],
                      "n_estimators": [100, 200, 500, 1000]}
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
#For KNN
grid_knn=GridSearchCV(estimator=knn,param_grid=params_knn,cv=5,n_jobs=-1)
grid_knn.fit(X_train,y_train)     
y_pred_grid_knn=grid_knn.predict(X_test)
print(accuracy_score(y_test,y_pred_grid_knn))
print(classification_report(y_test,y_pred_grid_knn))
print(confusion_matrix(y_test,y_pred_grid_knn))
```

```python
#For SVC
grid_svc=GridSearchCV(estimator=svc,param_grid=params_svc,cv=5,n_jobs=-1)
grid_svc.fit(X_train,y_train)     
y_pred_grid_svc=grid_svc.predict(X_test)
print(accuracy_score(y_test,y_pred_grid_svc))
print(classification_report(y_test,y_pred_grid_svc))
print(confusion_matrix(y_test,y_pred_grid_svc))
```

```python
#for decision tree
grid_tree=GridSearchCV(estimator=tree,param_grid=params_decision,n_jobs=-1)
grid_tree.fit(X_train,y_train)     
y_pred_grid_tree=grid_tree.predict(X_test)
print(accuracy_score(y_test,y_pred_grid_tree))
print(classification_report(y_test,y_pred_grid_tree))
print(confusion_matrix(y_test,y_pred_grid_tree))
```

```python
#for random forest
grid_random=RandomizedSearchCV(estimator=randomforest,param_distributions=params_random_forest,n_iter=10,cv=3,verbose=2,n_jobs=-1)
grid_random.fit(X_train,y_train)     
y_pred_grid_random=grid_random.predict(X_test)
print(accuracy_score(y_test,y_pred_grid_random))
print(classification_report(y_test,y_pred_grid_random))
print(confusion_matrix(y_test,y_pred_grid_random))
```
