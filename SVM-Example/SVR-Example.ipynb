```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("insurance.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
X=df.drop("charges",axis=1)
y=df["charges"]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=15)
```

```python
from sklearn.preprocessing import LabelEncoder,StandardScaler
encode=LabelEncoder()
X_train["sex"]=encode.fit_transform(X_train["sex"])
X_test["sex"]=encode.transform(X_test["sex"])
X_train["smoker"]=encode.fit_transform(X_train["smoker"])
X_test["smoker"]=encode.transform(X_test["smoker"])
X_train["region"]=encode.fit_transform(X_train["region"])
X_test["region"]=encode.transform(X_test["region"])
```

```python
X_train
```

```python
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
X_train
```

```python
from sklearn.linear_model import LinearRegression
linear=LinearRegression()
linear.fit(X_train,y_train)
y_pred_lin=linear.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
print("mse:",mean_squared_error(y_test,y_pred_lin))
print("mae:",mean_absolute_error(y_test,y_pred_lin))
print("score:",r2_score(y_test,y_pred_lin))
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly=PolynomialFeatures(degree=2)
X_train_poly=poly.fit_transform(X_train)
X_test_poly=poly.transform(X_test)
```

```python
poly_regression=LinearRegression()
poly_regression.fit(X_train_poly,y_train)
y_pred_poly=poly_regression.predict(X_test_poly)
print("mse:",mean_squared_error(y_test,y_pred_poly))
print("mae:",mean_absolute_error(y_test,y_pred_poly))
print("score:",r2_score(y_test,y_pred_poly))
```

```python
#Ridge Lasso Elastic
from sklearn.linear_model import RidgeCV,LassoCV,ElasticNetCV
ridgecv=RidgeCV(cv=5)
ridgecv.fit(X_train,y_train)
y_pred_ridge=ridgecv.predict(X_test)
print("mse:",mean_squared_error(y_test,y_pred_ridge))
print("mae:",mean_absolute_error(y_test,y_pred_ridge))
print("score:",r2_score(y_test,y_pred_ridge))
```

```python
lassocv=LassoCV(cv=5)
lassocv.fit(X_train,y_train)
y_pred_lasso=lassocv.predict(X_test)
print("mse:",mean_squared_error(y_test,y_pred_lasso))
print("mae:",mean_absolute_error(y_test,y_pred_lasso))
print("score:",r2_score(y_test,y_pred_lasso))
```

```python
elasticcv=ElasticNetCV(cv=5)
elasticcv.fit(X_train,y_train)
y_pred_elastic=elasticcv.predict(X_test)
print("mse:",mean_squared_error(y_test,y_pred_elastic))
print("mae:",mean_absolute_error(y_test,y_pred_elastic))
print("score:",r2_score(y_test,y_pred_elastic))
```

```python
#SVR
from sklearn.svm import SVR
svr=SVR()
svr.fit(X_train,y_train)
```

```python
y_pred_svr=svr.predict(X_test)
print("mse:",mean_squared_error(y_test,y_pred_svr))
print("mae:",mean_absolute_error(y_test,y_pred_svr))
print("score:",r2_score(y_test,y_pred_svr))
```
