```python
#Dependent and Independent
X=df.drop("price",axis=1)
y=df["price"]
```

```python
#Train test split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=15)
```

```python
X_train
```

```python
X_train=pd.get_dummies(X_train,columns=["CarName","fueltype","aspiration","doornumber","carbody",
                                        "drivewheel","enginelocation","enginetype","cylindernumber","fuelsystem"],drop_first=True,dtype=int)
```

```python
X_train
```

```python
X_test=pd.get_dummies(X_test,columns=["CarName","fueltype","aspiration","doornumber","carbody",
                                        "drivewheel","enginelocation","enginetype","cylindernumber","fuelsystem"],drop_first=True,dtype=int)
```

```python
X_test
```

```python
#For missing value
X_test = X_test.reindex(columns = X_train.columns, fill_value=0)
```

```python
#Standart Scaled
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
#Linear Regression
from sklearn.linear_model import LinearRegression
linear=LinearRegression()
linear.fit(X_train,y_train)
y_pred=linear.predict(X_test)
```

```python
#metrics
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
print("mae:",mean_absolute_error(y_test,y_pred))
print("mse:",mean_squared_error(y_test,y_pred))
print("Score:",r2_score(y_test,y_pred))
```

```python
#poly Regression
from sklearn.preprocessing import PolynomialFeatures
poly=PolynomialFeatures(degree=2)
X_train_poly=poly.fit_transform(X_train)
X_test_poly=poly.transform(X_test)
poly_linear=LinearRegression()
poly_linear.fit(X_train_poly,y_train)
y_pred_poly=poly_linear.predict(X_test_poly)
print("mae:",mean_absolute_error(y_test,y_pred_poly))
print("mse:",mean_squared_error(y_test,y_pred_poly))
print("Score:",r2_score(y_test,y_pred_poly))
#Feature explosion: creating too many large-valued polynomial features, causing model predictions to blow u
```

```python
#SVR
from sklearn.svm import SVR
svr=SVR()
svr.fit(X_train,y_train)
y_pred_svr=svr.predict(X_test)
print("mae:",mean_absolute_error(y_test,y_pred_svr))
print("mse:",mean_squared_error(y_test,y_pred_svr))
print("Score:",r2_score(y_test,y_pred_svr))
```

```python
#hyperparameter SVR
params={"kernel":["linear","poly","rbf","sigmoid"],
        "C":[0.01,0.1,1,10,100,1000],
        "gamma":["scale", "auto"]}
```

```python
#GridSearch
from sklearn.model_selection import GridSearchCV
grid=GridSearchCV(estimator=svr,param_grid=params,cv=5,n_jobs=-1)
grid.fit(X_train,y_train)
y_pred_svhype=grid.predict(X_test)
print("mae:",mean_absolute_error(y_test,y_pred_svhype))
print("mse:",mean_squared_error(y_test,y_pred_svhype))
print("Score:",r2_score(y_test,y_pred_svhype))
```
