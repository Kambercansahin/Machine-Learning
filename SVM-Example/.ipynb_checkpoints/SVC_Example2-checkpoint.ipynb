```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as snsx
```

```python
df.drop(["id","Unnamed: 32"],axis=1,inplace=True)
```

```python
df.info()
```

```python
df.head()
```

```python
#sns.pairplot(df)
```

```python
X=df.drop("diagnosis",axis=1)
y=df["diagnosis"]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=15)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label=LabelEncoder()
y_train=label.fit_transform(y_train)
y_test=label.transform(y_test)
```

```python
sns.boxplot(data=X_train)
```

```python
#We have a lot of outlier in dataset
from sklearn.preprocessing import RobustScaler
Rscaler=RobustScaler()
X_train=Rscaler.fit_transform(X_train)
X_test=Rscaler.transform(X_test)
```

```python
sns.boxplot(data=X_train)
```

```python
from sklearn.linear_model import LogisticRegression
logistic=LogisticRegression(max_iter=5000)
logistic.fit(X_train,y_train)
```

```python
y_pred=logistic.predict(X_test)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```

```python
#Hyperparameter
parameter={"C":[0.1,1,10,100,1000],
           "penalty":["l1", "l2", "elasticnet"],
           "solver":["lbfgs", "liblinear", "newton-cg", "newton-cholesky", "sag", "saga"],
           "class_weight":[{0:w,1:y} for w in [1,10,100] for y in [1,10,100]]
           }
            
```

```python
from sklearn.model_selection import GridSearchCV,StratifiedKFold
cv=StratifiedKFold()
grid_logistic=GridSearchCV(estimator=logistic,param_grid=parameter,cv=cv,n_jobs=-1)
grid_logistic.fit(X_train,y_train)
y_pred_logistic=grid_logistic.predict(X_test)
print(classification_report(y_test,y_pred_logistic))
print(confusion_matrix(y_test,y_pred_logistic))
```

```python
#ROC and AUC
from sklearn.metrics import roc_auc_score,roc_curve
model_prob=grid_logistic.predict_proba(X_test)
```

```python
model_prob=model_prob[:,1]
```

```python
model_auc=roc_auc_score(y_test,model_prob)
```

```python
model_auc
```

```python
#SVC
from sklearn.svm import SVC
svc=SVC(kernel="rbf")
svc.fit(X_train,y_train)
```

```python
y_pred_svc=svc.predict(X_test)
print(classification_report(y_test,y_pred_svc))
print(confusion_matrix(y_test,y_pred_svc))
```

```python
#SVC hyperparameter
params={"C":[0.1,1,10,100,1000],
        "kernel":["linear", "poly", "rbf", "sigmoid"],
        "gamma":["scale","auto"]}
```

```python
cv=StratifiedKFold()
grid_svc=GridSearchCV(estimator=svc,param_grid=params,cv=cv,n_jobs=-1)
grid_svc.fit(X_train,y_train)
y_pred_svc=svc.predict(X_test)
print(classification_report(y_test,y_pred_svc))
print(confusion_matrix(y_test,y_pred_svc))
```
