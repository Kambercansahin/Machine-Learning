```python
#Dependent independent features
X=df.drop("income",axis=1)
y=df["income"]
```

```python
#Train test split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=15)
```

```python
#Encode 
X_train=pd.get_dummies(X_train,columns=["workclass","relationship","occupation","marital.status","sex","race","native.country"],drop_first=True,dtype=int)
X_test=pd.get_dummies(X_test,columns=["workclass","relationship","occupation","marital.status","sex","race","native.country"],drop_first=True,dtype=int)


```

```python
X_test
```

```python
X_train
```

```python
#for missing value columns
X_test=X_test.reindex(columns=X_train.columns,fill_value=0)
```

```python
#Scaler
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
#Encode for y
from sklearn.preprocessing import LabelEncoder
label=LabelEncoder()
y_train=label.fit_transform(y_train)
y_test=label.transform(y_test)
```

```python
#Logistic regression
from sklearn.linear_model import LogisticRegression
logistic=LogisticRegression()
logistic.fit(X_train,y_train)
```

```python
y_pred=logistic.predict(X_test)
from sklearn.metrics import classification_report,confusion_matrix
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```

```python
#hyperparameter logistic
params={"penalty":["l1", "l2","elasticnet"],
        "C":[0.01,0.1,1,10,100,1000],
        "solver":["lbfgs","liblinear","newton-cg","newton-cholesky","sag","saga"]}

from sklearn.model_selection import GridSearchCV,StratifiedKFold
cv=StratifiedKFold()
grid=GridSearchCV(estimator=logistic,cv=cv,param_grid=params,n_jobs=-1)
grid.fit(X_train,y_train)
```

```python
y_pred_grid=grid.predict(X_test)
print(classification_report(y_test,y_pred_grid))
print(confusion_matrix(y_test,y_pred_grid))
```

```python
#SVC
from sklearn.svm import SVC
svc=SVC()
svc.fit(X_train,y_train)
y_pred_svc=svc.predict(X_test)
print(classification_report(y_test,y_pred_svc))
print(confusion_matrix(y_test,y_pred_svc))
```

```python
#hyperparam svc
params_svc={"C":[0.01,0.1,1,10,50,100,1000],
            "kernel":["linear","poly","rbf","sigmoid"],
            "gamma":["scale","auto"]}
grid_svc=GridSearchCV(estimator=svc,cv=cv,param_grid=params_svc,n_jobs=-1)
grid_svc.fit(X_train,y_train)
```

```python
y_pred_svch=grid_svc.predict(X_test)
print(confusion_matrix(y_test,y_pred_svch))
print(classification_report(y_test,y_pred_svch))
```

```python
#KNN
from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier()
knn.fit(X_train,y_train)
```

```python
y_pred_knn=knn.predict(X_test)
print(confusion_matrix(y_test,y_pred_knn))
print(classification_report(y_test,y_pred_knn))
```

```python
#Navie Bayes
from sklearn.naive_bayes import GaussianNB,MultinomialNB
gnb=GaussianNB()
multi=MultinomialNB()

gnb.fit(X_train,y_train)
```

```python
y_pred_gnb=gnb.predict(X_test)
print(confusion_matrix(y_test,y_pred_gnb))
print(classification_report(y_test,y_pred_gnb))
```
